Other algorithms were discussed in the lab, however this was chosen as the main example as it allows for a demonstration of how algorithms and pseudocodes can be used in humanities research. Both of these are highly relevant in Digital Humanities research as it presents the tasks that will be done by researchers that will be carrying out the code in a manner that can be easily understood by those who do not understand code. Translating algorithms into pseudocode provides further detail into exactly what is being done by the chosen programming language to the text. This is beneficial for the coder as it can serve as a detailed plan for the code that needs to be written out. It is also beneficial for colleagues working with the coder as it provides specifications into what is being done to the text. In the event that the actual code cannot be shared, having the programming language, software used, and the pseudocode will help demonstrate both aspects. The example below shows how the algorithm above can be translated into a pseudocode:

**Pseudocode: Finding the number of times women have been mentioned in a text.**

    INPUT: chapbook text

    split text into list of words and assign it to a variable called "words"

    created an empty dictionary and assign it to a variable called "counts"

    create list of keywords related to women in the example text

    FOR each word in counts:
        IF word is already in the dictionary counts:
            increase the current count of the word by one
            assign the new value to the entry for the word in the dictionary

        ELSE:
            assign the value 1 to the entry for the word in the dictionary
    
    FIND how many times each keyword was mentioned

    OUTPUT counts for each keyword

This pseudocode was then submitted to Microsoft Copilot in order to identify possible errors. This ensures that future amendments to the code can be made without too many additional changes.

**Corrected pseudocode provided by Microsoft Copilot:**
