Task 3 involves shuffling and stratifying the data, creating a set of test data from the entire data set which can be used to create and train the ensuing Machine Learning algorithm. This task involves shuffling the data, carrying out stratified sampling for data that has not been cleaned, and setting aside test data for datasets that have already been cleaned. **Evidence** In the stratified sampling example, it demonstrates how to retrieve a truly randomised number of data points from the dataset so that the results from the machine learning algorithm produce outputs that are less biased, if not totally unbiased. It also contributes to the accuracy of the machine learning algorithm, allowing it to produce outputs that are more representative of the population that they are trained on. Given how the outputs of existing Machine Learning algorithms are often taken at face-value, it is important that this step is carried out properly such that the chance of being given an inaccurate output is minimal.
