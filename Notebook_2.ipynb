Task 3 involves shuffling and stratifying the data, creating a set of test data from the entire data set which can be used to create and train the ensuing Machine Learning algorithm. It also contributes to the accuracy of the machine learning algorithm, allowing it to produce outputs that are more representative of the population that they are trained on.
